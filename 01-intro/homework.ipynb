```python
asia = df[df["origin"].str.strip().str.lower() == "asia"]

print(asia["fuel_efficiency_mpg"].max())
```

```python
median_before = df["horsepower"].median()
most_frequent = df["horsepower"].mode().iloc[0]

horsepower_filled = df["horsepower"].fillna(most_frequent)
median_after = horsepower_filled.median()

print("Most frequent horsepower:", most_frequent)
print("Median before:", median_before)
print("Median after:", median_after)
```

```python
import numpy as np

X = asia[["vehicle_weight", "model_year"]].head(7).to_numpy()

XTX = X.T @ X
XTX_inverse = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inverse @ X.T @ y

print("Weights:", w)
print("Sum of weights:", w.sum())
```
